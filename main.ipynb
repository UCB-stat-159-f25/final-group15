# Bay Wheels ridership and weather in San Francisco, 2018 to 2024

## Motivation
I want to understand how daily weather relates to daily Bay Wheels ridership in San Francisco. The city has a mild climate, so it is interesting that even small changes in temperature or rain might change how many people ride.

## Data and source
Bike trips come from monthly Bay Wheels trip files that we aggregated to the day level. Weather comes from NOAA daily summaries for station USW00023272 in San Francisco. We merged bike daily totals with daily weather on the date and kept 2018 to 2024. The merged dataset has 2528 daily observations. (see cleaner.ipynb)


## Research questions
1. Do warmer days have more bike trips
2. Do rainy days have fewer bike trips
3. Is there a clear seasonal pattern across months and years

## Methods and assumptions
This is an observational analysis. We assume the station weather is a reasonable proxy for city conditions and that the date merge is correct. We treat missing or special coded weather values as missing during cleaning. We use correlation plots and simple regressions with robust standard errors. We also add day of week controls to reduce obvious calendar confounding. (see cleaner.ipynb)

## Results

### Question 1
Temperature is positively related to ridership. In the merged data, average temperature has a moderate positive correlation with total trips at about 0.36. (see visualization_analysis.ipynb)
In regression, a 1 degree C increase in average temperature is associated with about 270 more trips per day, and the estimate is statistically significant with robust standard errors.(see visualization_analysis.ipynb)
Even though most days are between about 10 and 17 C, the scatterplot still shows an upward trend, which is a nice reminder that small comfort changes matter in SF. (see visualization_analysis.ipynb)

### Question 2
Rain reduces ridership. Precipitation has a negative correlation with total trips around minus 0.28.(see visualization_analysis.ipynb)

In regression, 1 mm more rain is associated with about 90 fewer trips per day.(see visualization_analysis.ipynb)

If we instead use a rainy day indicator, rainy days have about 1480 fewer trips on average, even after controlling for day of week.(see visualization_analysis.ipynb)

The boxplot also shows a big gap between rainy and dry days, which matches the model. (see visualization_analysis.ipynb)


### Question 3
Ridership follows a seasonal pattern. Average trips rise through spring and summer, peak around September to October, then fall in winter. (see visualization_analysis.ipynb)

When we split by precipitation buckets, dry days consistently have the highest usage and trips generally increase from 2018 to 2024, with a dip around 2021. (see visualization_analysis.ipynb)


## Limitations
This is not a causal claim. Weather is not randomly assigned and other factors like commuting patterns, holidays, pricing, station availability, and special events can move ridership. Our simple controls help a bit, but they do not remove all confounding. The model R squared is about 0.17, so a lot of variation is still explained by factors outside temperature and rain. (see visualization_analysis.ipynb)


## Conclusion
Even in San Francisco’s mild climate, warmer and drier days have meaningfully higher bike usage. Rain has the largest immediate impact and can reduce trips by about 1500 on a typical rainy day.
